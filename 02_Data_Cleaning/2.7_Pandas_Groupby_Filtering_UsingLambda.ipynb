## 🧮 1️⃣ Salary Normalization Using GroupBy and Transform

### 💡 Code
```python
df['salary_norm'] = df['salary'] / df.groupby('city')['salary'].transform('mean')
```

---

### 🎯 Purpose
This line of code **normalizes the salary of each person relative to the average salary in their city**.

In other words:
> It creates a new column called `salary_norm` that shows how a person’s salary compares to their city’s average.

- `salary_norm = 1` → salary is **equal to** the city’s average  
- `salary_norm > 1` → salary is **above** the city’s average  
- `salary_norm < 1` → salary is **below** the city’s average

---

### 🔍 Step-by-Step Breakdown

#### 🧩 Step 1: Grouping the Data
```python
df.groupby('city')['salary']
```
This groups the DataFrame by `city` and focuses on the `salary` column.

If the DataFrame is:

| name | city | salary |
|------|------|--------|
| A | Delhi | 50000 |
| B | Delhi | 100000 |
| C | Mumbai | 70000 |
| D | Mumbai | 140000 |

Then `groupby('city')['salary']` creates:
- Delhi → [50000, 100000]
- Mumbai → [70000, 140000]

---

#### 🧮 Step 2: Calculating the Mean with `.transform('mean')`
```python
df.groupby('city')['salary'].transform('mean')
```
This calculates the **mean salary for each city**, and **returns a Series of the same length as the original DataFrame**.

| name | city | salary | mean_salary |
|------|------|--------|-------------|
| A | Delhi | 50000 | 75000 |
| B | Delhi | 100000 | 75000 |
| C | Mumbai | 70000 | 105000 |
| D | Mumbai | 140000 | 105000 |

**Note:** `.transform('mean')` keeps the same number of rows as the original DataFrame, repeating the group mean for each corresponding row.  
This is what makes it different from `.mean()` which would only return one mean per group.

---

#### ➗ Step 3: Performing the Division
Now, divide each person’s salary by their city’s mean:

| name | city | salary | mean_salary | salary_norm = salary / mean |
|------|------|--------|-------------|-----------------------------|
| A | Delhi | 50000 | 75000 | 0.67 |
| B | Delhi | 100000 | 75000 | 1.33 |
| C | Mumbai | 70000 | 105000 | 0.67 |
| D | Mumbai | 140000 | 105000 | 1.33 |

---

#### 🆕 Step 4: Assigning to a New Column
```python
df['salary_norm'] = ...
```
This stores the result as a new column `salary_norm`.

Final DataFrame 👇

| name | city | salary | salary_norm |
|------|------|--------|-------------|
| A | Delhi | 50000 | 0.67 |
| B | Delhi | 100000 | 1.33 |
| C | Mumbai | 70000 | 0.67 |
| D | Mumbai | 140000 | 1.33 |

---

### 🧠 Intuitive Understanding
This tells us **how much above or below the average** a salary is within its city.

For example:
- B earns **33% above** the Delhi average.
- A earns **33% below** the Delhi average.
- C earns **33% below** the Mumbai average.
- D earns **33% above** the Mumbai average.

---

### ⚙️ Why Use `.transform()` Instead of `.mean()`?
If we used `.mean()`, we’d get only one mean value per city, not a value per row.

✅ `.mean()` → Returns one value per group.  
✅ `.transform('mean')` → Returns one value per row (matching original index).

That’s why `.transform()` is essential for **row-wise normalization within groups**.

* **Format 1 `groupby(["col1", "col2"])`**: Creates groups based on **combinations**.

* **Format 2 `groupby('col1')['col2']`**: Creates groups based on **one column**, then "selects" a **different column** from those groups to perform a calculation on.

## Format 1: Grouping by Multiple Columns

```python
ndf.groupby(['year_added', 'type'])
```

* **What it means:** "Create groups from every unique **combination** of `year_added` AND `type`."

* **Analogy:** Think of sorting laundry. This is like making piles for "Red Shirts," "Red Socks," "Blue Shirts," "Blue Socks." You are grouping by *both* color and item type at the same time.

* **Resulting Groups (Conceptually):**

  * Group 1: (2019, Movie)

  * Group 2: (2019, TV Show)

  * Group 3: (2020, Movie)

  * Group 4: (2020, TV Show)

  * ...and so on.

* **When you use it:** You do this when your calculation needs to know *both* values. Your previous code, `.size()`, was a perfect example. You wanted to know the size of the "(2019, Movie)" pile, not just the "2019" pile.

## Format 2: Grouping by One, Aggregating Another

```python
df.groupby('col1')['col2']
```

* **What it means:** "Group the rows *only* by `col1`. Then, from each of those groups, grab *only* the values from `col2`."

* **Analogy:** Using the same laundry example, this is like making piles for "Reds" and "Blues" (grouping by `col1` = color). Then, from the "Reds" pile, you *only* grab the socks (`col2` = item type) to count them. You ignore the red shirts.

* **Example:** Let's find the *average duration* for Movies vs. TV Shows.

```python

# 1. Group by 'type' (Movie, TV Show)
# 2. Grab the 'duration' column from each group
# 3. Calculate the mean (average) of those durations
df.groupby('type')['duration'].mean()

```

* **When you use it:** This is the most common format when you want to ask a question like, "What is the **\[average/total/max/min\] \[numeric column\]** for each **\[category\]**?"

### Summary Cheat Sheet

| **Feature** | **Format 1: groupby(\["col1", "col2"\])** | **Format 2: groupby('col1')\['col2'\]** | 
| **Purpose** | Get a single statistic (like count) for a *combined* group. | Get a statistic (like average) from a *specific column* for a group. | 
| **Question it Answers** | "How many items are in the (A, X) category?" | "What is the average of *Column Y* for all items in category A?" | 
| **Your Example** | `ndf.groupby(['year', 'type']).size()` | `ndf.groupby('type')['duration'].mean()` | 
| **Group By...** | *Multiple* columns (`col1` AND `col2`) | *One* column (`col1`) | 
| **Operates On...** | The whole group | *Only* `col2` within that group |

## 💡 2️⃣ Filtering Rows Using Lambda in `.loc[]`

### 🧾 Code
```python
df.loc[lambda d: d['age'] > 25, :]
```

---

### 🎯 Purpose
Select all rows from the DataFrame where the `age` column is greater than 25.

Equivalent to:
```python
df[df['age'] > 25]
```
But the lambda version is **cleaner and more flexible** when used in method chaining.

---

### 🧩 Step-by-Step Breakdown

#### 🪄 Step 1: `.loc[]`
`.loc[]` is a **label-based indexer** in Pandas.

Basic format:
```python
df.loc[row_condition, column_selection]
```

- Before the comma → select rows (boolean condition).  
- After the comma → select columns.

Example:
```python
df.loc[df['age'] > 25, ['name', 'city']]
```
→ Select rows where `age > 25`, and display only `name` and `city`.

---

#### 🧠 Step 2: The Lambda Function
`lambda d: d['age'] > 25` defines an **anonymous inline function** that takes the entire DataFrame (`d`) and returns a **boolean Series**.

It’s equivalent to:
```python
def filter_condition(d):
    return d['age'] > 25
```

So for a DataFrame like:

| name | age | city |
|------|-----|------|
| A | 22 | Delhi |
| B | 30 | Mumbai |
| C | 28 | Chennai |
| D | 24 | Pune |

`d['age'] > 25` gives:

| name | age | city | Condition |
|------|-----|------|------------|
| A | 22 | Delhi | False |
| B | 30 | Mumbai | True |
| C | 28 | Chennai | True |
| D | 24 | Pune | False |

---

#### ⚙️ Step 3: How `.loc` Uses It
`.loc` uses this boolean mask to **filter rows** where the condition is True.

```python
df.loc[lambda d: d['age'] > 25, :]
```

The `:` means **select all columns**.

Output:

| name | age | city |
|------|-----|------|
| B | 30 | Mumbai |
| C | 28 | Chennai |

---

### 🧩 Step 4: Why Use Lambda Here?
You could write the same logic as:
```python
df.loc[df['age'] > 25, :]
```
But using lambda allows **cleaner chaining**, such as:

```python
df.loc[lambda d: d['age'] > 25]   .assign(status='Adult')   .sort_values('age', ascending=False)
```

This makes your Pandas pipeline more **functional and elegant** — ideal for clean data analysis scripts.

---

### ✅ Summary Table

| Part | Meaning |
|------|----------|
| `df.loc[...]` | Select rows and columns using labels |
| `lambda d:` | Define an inline function using the DataFrame |
| `d['age'] > 25` | Returns a boolean mask |
| `, :` | Select all columns |
| **Result** | Keeps only rows where `age > 25` |

---

### ✨ Example Code
```python
import pandas as pd

df = pd.DataFrame({
    'name': ['A', 'B', 'C', 'D'],
    'age': [22, 30, 28, 24],
    'city': ['Delhi', 'Mumbai', 'Chennai', 'Pune']
})

result = df.loc[lambda d: d['age'] > 25, :]
print(result)
```

**Output:**
```
  name  age    city
1    B   30  Mumbai
2    C   28  Chennai
```

---

### ⚡ Alternate Version
```python
df.loc[df['age'] > 25, :]
```
Both are identical in result.  
Lambda version is just more expressive when chaining multiple operations.
